##ESTRUCTURA DE REPETECION
PERMITE REPETIR UNA ACCION N CANTIDAD DE VECES O HASTA QUE LOGRE UNA CONDICION

##ESTRUCTURA PARA (FOR)
Empleada cuando se sabe cuantas veces se debe repetir una accion

```python
for intento in range (31):
  if intento%3==0:
    print(intento)
```

```python
for prueba in range (3,31,3):
  print(prueba)
```

```python
n=int(input("ponga un numero:"))
for tabla in range(1,11):
  print(n,"x", n, "=", n*tabla)
```

```python
for n in range(1,11):
  print("tabla del",n)
  for i in range(1,11):
    print(n,"x", i,"=",n*i)
    print()
```
