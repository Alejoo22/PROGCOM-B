```python
total_personas = 5
constructor=[]

for i in range(total_personas):
 exp = int(input(f"Ingrese la experiencia de la persona {i}: "))
 constructor.append(exp)

```

```python
promedio=sum(constructor)/len(constructor)
promedio
for i in range(total_personas):
  if constructor[i]>promedio :
    print(f"el constructor{i} tiene mas experiencia que el promedio")
```
