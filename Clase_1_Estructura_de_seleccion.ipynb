#Estructura de seleccion


##Estructura if/else
evalua una condicion y da un valor de verdad:
(V) o (F)
sintaxis
```
if (condicione):
 se ejecutan operaciones si condicion es verdadera
else:
 se ejecutan operaciones si condicion es falsa
```



##if/elif/else
evalua tantas condiciones como elif hayan. Si no se  cumple, pasa a la siguiente condicion, hasta llegar al else

sintaxis
```
if (condicion):
aqui se ejecuta el codigo si condicion1 es V
elif (condicion2):
se ejecuta si condicion1 es falsa pero condicion2 es verdadera
else (ninguna de las anteriores)
```



##match/case
la estructura match evalua una variable frente a avarios bloques case. el comodin case__ funciona como ninguna de las anteriores.

sintaxis
```
match(condicion):
 case v1: aqui se ejecuta si condicion==v1
 case v2: aqui se ejecuta si condicion==v2
 case v3: aqui se ejecuta si condicion==v3
 case__: ninguna de las anteriores
```



````python
helado=input("quieres helado? ")
````

````python
#if/else
if helado.lower()=="si"or helado.lower()=="si": print("pos te lo compro")
else: print("pos no te lo compro")
````

````python
#if/elif/else
if helado.lower()=="si":print("pos te lo compro")
elif helado.lower()=="si":print("pos te lo compro")
elif helado.lower()=="sipi":print("pos te lo compro")
else:print("Pos no te lo compro")
````

````python
#match/case
match helado.lower():
  case "si":print("Pos te lo compro")
  case "si":print("Pos te lo compro")
  case "sipis":print("Pos te lo compro")
  case "clarinetes":print("Pos te lo compro")
  case __:print("Pos no te lo compro")
````
