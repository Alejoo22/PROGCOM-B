```python
flores = {
    "tipo": ["rosa", "tulipan", "girasol", "amapola", "lirio"],
    "cant": [10, 17, 3, 8, 5],
    "precios": [4000, 80000, 5000, 15000, 23000]
}

compra_user = {
    "tipo": [],
    "cantidad": [],
    "precio": []
}

IVA = 0.19
total_pagar = 0

print("Bienvenido a mi floristeria - Precios sin IVA")

while True:
    if len(flores["tipo"]) == 0:
        print("\n¡Ya no hay flores en inventario! Saliendo...")
        break

    print("\n¿Qué desea hacer?")
    print("1. Comprar flores")
    print("2. Pagar")
    print("3. Generar factura con IVA")
    print("4. Quitar del carrito")
    print("5. Salir")

    opc = input("Seleccione su opción:\n")

    match opc:
        case "1":
            print("Ha seleccionado: Comprar flores\n")
            print("--- STOCK ACTUAL - Precio sin IVA ---")

            for t, c, p in zip(
                flores["tipo"],
                flores["cant"],
                flores["precios"]
            ):
                print(f"{t:<10} | Cant: {c} | ${p} + IVA")

            print("--------------------\n")

            flor_opc = input(
                "Seleccione la flor que desea comprar: "
            ).lower().strip()

            if flor_opc in flores["tipo"]:
                idx = flores["tipo"].index(flor_opc)

                flor_cant = int(
                    input("Ingrese la cantidad que desea comprar: ")
                )

                if 0 < flor_cant <= flores["cant"][idx]:
                    compra_user["tipo"].append(flor_opc)
                    compra_user["cantidad"].append(flor_cant)
                    compra_user["precio"].append(
                        flores["precios"][idx]
                    )

                    flores["cant"][idx] -= flor_cant

                    print(
                        f"¡Se han añadido {flor_cant} "
                        f"{flor_opc}(s) al carrito!"
                    )

                    if flores["cant"][idx] == 0:
                        flores["tipo"].pop(idx)
                        flores["cant"].pop(idx)
                        flores["precios"].pop(idx)
                else:
                    print("Cantidad no disponible")
            else:
                print(f"La flor {flor_opc} no está disponible")

        case "2":
            if len(compra_user["tipo"]) == 0:
                print("El carrito de compras está vacío.")
            else:
                print("\n*** TABLA DE COMPRAS ***")

                subtotal_sin_iva = 0

                for tipo, cantidad, precio in zip(
                    compra_user["tipo"],
                    compra_user["cantidad"],
                    compra_user["precio"]
                ):
                    sub = cantidad * precio
                    subtotal_sin_iva += sub

                iva_valor = subtotal_sin_iva * IVA
                total_pagar = subtotal_sin_iva + iva_valor

                print(
                    f"{'Flores':<10} | {'Cant':<5} | "
                    f"{'Unit':<8} | {'Subtotal'}"
                )
                print("-" * 55)

                for tipo, cantidad, precio in zip(
                    compra_user["tipo"],
                    compra_user["cantidad"],
                    compra_user["precio"]
                ):
                    print(
                        f"{tipo.capitalize():<10} | "
                        f"{cantidad:<5} | "
                        f"${precio:<7} | "
                        f"${cantidad * precio}"
                    )

                print("-" * 55)
                print(f"Subtotal sin IVA: ${subtotal_sin_iva}")
                print(f"IVA ({IVA * 100:.0f}%): ${iva_valor:.0f}")
                print(f"TOTAL A PAGAR: ${total_pagar:.0f}\n")

                print("Seleccione el medio de pago:")
                print("1. Efectivo")
                print("2. Tarjeta de crédito")
                print("3. Tarjeta de débito")

                pago_opc = input("Seleccione su opción:\n")

                if pago_opc == "1" or pago_opc == "3":
                    plata = int(
                        input(
                            f"El total con IVA es ${total_pagar:.0f}. "
                            "¿Cuánto dinero tiene? "
                        )
                    )

                    if plata >= total_pagar:
                        print(
                            f"Pago exitoso. Cambio: "
                            f"${plata - total_pagar:.0f}"
                        )
                    else:
                        print(
                            f"¡No le alcanza! Tiene ${plata} "
                            f"y necesita ${total_pagar:.0f}"
                        )

                elif pago_opc == "2":
                    cuotas = int(
                        input("¿Número de cobros / cuotas? ")
                    )

                    if cuotas > 0:
                        print(
                            f"Se cobrará en {cuotas} cuotas "
                            f"de ${total_pagar / cuotas:.0f}"
                        )
                    else:
                        print("El número de cuotas debe ser mayor que cero.")

                else:
                    print("Medio de pago incorrecto.")

        case "3":
            print("Ha seleccionado: Generar factura con IVA\n")

            if len(compra_user["tipo"]) == 0:
                print("El carrito de compras está vacío.")
            else:
                subtotal_sin_iva = sum(
                    c * p
                    for c, p in zip(
                        compra_user["cantidad"],
                        compra_user["precio"]
                    )
                )

                iva_valor = subtotal_sin_iva * IVA
                total_con_iva = subtotal_sin_iva + iva_valor

                print(
                    f"{'Flores':<12} | {'Cantidad':<8} | "
                    f"{'Unit':<8} | {'Subtotal'}"
                )
                print("-" * 60)

                for tipo, cantidad, precio in zip(
                    compra_user["tipo"],
                    compra_user["cantidad"],
                    compra_user["precio"]
                ):
                    print(
                        f"{tipo.capitalize():<12} | "
                        f"{cantidad:<8} | "
                        f"${precio:<7} | "
                        f"${precio * cantidad}"
                    )

                print("-" * 60)
                print(f"SUBTOTAL: ${subtotal_sin_iva}")
                print(f"IVA 19%: ${iva_valor:.0f}")
                print(f"TOTAL FACTURA: ${total_con_iva:.0f}")

        case "4":
            print("Ha seleccionado: Quitar del carrito\n")

            if len(compra_user["tipo"]) == 0:
                print("Carrito vacío")
            else:
                for t, c in zip(
                    compra_user["tipo"],
                    compra_user["cantidad"]
                ):
                    print(f"- {t}: {c}")

                quitar_flor = input(
                    "¿Qué flor desea quitar? "
                ).lower().strip()

                if quitar_flor in compra_user["tipo"]:
                    idx_c = compra_user["tipo"].index(quitar_flor)
                    cant_comprada = compra_user["cantidad"][idx_c]

                    quitar_cant = int(
                        input(
                            f"Usted compró {cant_comprada} "
                            f"{quitar_flor}(s). "
                            "¿Cuántas desea quitar? "
                        )
                    )

                    if quitar_cant <= 0:
                        print("La cantidad debe ser mayor que cero.")

                    elif quitar_cant > cant_comprada:
                        print(
                            f"Usted compró {cant_comprada}; "
                            f"no puede quitar más de esa cantidad."
                        )

                    else:
                        if quitar_flor in flores["tipo"]:
                            idx_f = flores["tipo"].index(quitar_flor)
                            flores["cant"][idx_f] += quitar_cant
                        else:
                            flores["tipo"].append(quitar_flor)
                            flores["cant"].append(quitar_cant)
                            flores["precios"].append(
                                compra_user["precio"][idx_c]
                            )

                        compra_user["cantidad"][idx_c] -= quitar_cant

                        if compra_user["cantidad"][idx_c] == 0:
                            compra_user["tipo"].pop(idx_c)
                            compra_user["cantidad"].pop(idx_c)
                            compra_user["precio"].pop(idx_c)

                        print(
                            f"Se retiró {quitar_cant} "
                            f"{quitar_flor}(s)"
                        )
                else:
                    print("Esa flor no está en el carrito.")

        case "5":
            print("Ha seleccionado: Salir\n")
            break

        case _:
            print("Número incorrecto")

print("Gracias por emplear nuestros servicios. Feliz día")
```
