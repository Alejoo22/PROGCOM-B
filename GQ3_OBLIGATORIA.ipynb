##PROBLEMA FINANCIERO
En finanzas personales, uno de los principales problemas no es la falta de información, sino la falta de constancia y motivación para cumplir metas de ahorro o inversión. Mi programa aplica gamificación igual que hacen apps financieras reales para que el usuario visualice su progreso como si fuera un videojuego, aumentando la probabilidad de que cumpla sus objetivos financieros.

```python
misiones_financieras = []

cantidad = int(input("Cuantas misiones financieras quieres ingresar? "))

for i in range(cantidad):
    print("\nMision", i + 1)
    nombre = input("Nombre de la mision: ")
    meta = float(input("Monto meta: "))
    actual = float(input("Monto actual ahorrado: "))

    exp = int(meta / 1000)

    mision = {"nombre": nombre, "monto_meta": meta, "monto_actual": actual, "exp": exp}
    misiones_financieras.append(mision)

exp_total = 0
misiones_completadas = []
misiones_pendientes = []

print("\n--- Resultado ---")

for mision in misiones_financieras:
    nombre = mision["nombre"]
    meta = mision["monto_meta"]
    actual = mision["monto_actual"]

    if actual >= meta:
        print(nombre, "-> COMPLETADA, exp ganada:", mision["exp"])
        exp_total = exp_total + mision["exp"]
        misiones_completadas.append(nombre)
    else:
        faltante = meta - actual
        progreso = (actual / meta) * 100
        print(nombre, "-> EN PROGRESO,", round(progreso, 1), "% completado, faltan", faltante)
        misiones_pendientes.append(nombre)

print("\nExperiencia financiera total:", exp_total)

if exp_total >= 3000:
    nivel = "Inversionista Experto"
elif exp_total >= 1500:
    nivel = "Ahorrador Avanzado"
elif exp_total >= 500:
    nivel = "Aprendiz de Finanzas"
else:
    nivel = "Novato Financiero"

print("Tu nivel actual es:", nivel)

resumen_completadas = [m["nombre"] for m in misiones_financieras if m["monto_actual"] >= m["monto_meta"]]
print("\nMisiones completadas:")
for nombre in resumen_completadas:
    print("-", nombre)

```
