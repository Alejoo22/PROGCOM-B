##GP1- PIEDRA, PAPEL O TIJERA
Decido hacer este quest obligatorio un poco distinto al resto, lo haré en inglés, fomentando un poco a aprender el idioma, ya que es fundamental manejar el idioma inglés en programación y en la vida.

```python
import random

opciones = ["rock", "paper", "scissors", "lizard", "spock"]

gana_a = {
    "rock": ["scissors", "lizard"],
    "paper": ["rock", "spock"],
    "scissors": ["paper", "lizard"],
    "lizard": ["spock", "paper"],
    "spock": ["scissors", "rock"]
}

ganadas = 0
perdidas = 0
empates = 0
historial = []

rondas = int(input("How many rounds do you want to play? "))

for i in range(rondas):
    print("Round", i + 1)
    jugador = input("Choose rock, paper, scissors, lizard or spock: ")

    if jugador not in opciones:
        print("Invalid choice")
        continue

    computadora = random.choice(opciones)
    print("Computer chose:", computadora)

    if jugador == computadora:
        print("It's a tie!")
        empates = empates + 1
    else:
        if computadora in gana_a[jugador]:
            print("You win!")
            ganadas = ganadas + 1
        else:
            print("You lose!")
            perdidas = perdidas + 1

    historial.append(jugador + " vs " + computadora)

print("Wins:", ganadas)
print("Losses:", perdidas)
print("Ties:", empates)

resumen = [j for j in historial]
print(resumen)
```
