##GP1- PIEDRA, PAPEL O TIJERA
Decido hacer este quest obligatorio un poco distinto al resto, lo haré en inglés, fomentando un poco a aprender el idioma, ya que es fundamental manejar el idioma inglés en programación y en la vida.

```python
import random

opciones = ["piedra", "papel", "tijera", "pistola", "trueno", "diablo", "dragon", "agua", "aire", "esponja", "lobo", "arbol", "humano", "serpiente", "fuego"]

n = len(opciones)
gana_a = {}

for i in range(n):
 elemento = opciones[i]
 rivales = [opciones[(i + j) % n] for j in range(1, 8)]
 gana_a[elemento] = rivales

ganadas = 0
perdidas = 0
empates = 0
historial = []

rondas = int(input("Cuantas rondas quieres jugar? "))

for i in range(rondas):
 print("\nRonda", i + 1)
 print("Opciones:", opciones)
 jugador = input("Elige tu jugada: ")

 if jugador not in opciones:
  print("Opcion invalida")
  continue

 computadora = random.choice(opciones)
 print("La computadora eligio:", computadora)

 if jugador == computadora:
  print("Empate2")
  empates = empates + 1
 else:
   if computadora in gana_a[jugador]:
    print("Ganaste")
    ganadas = ganadas + 1
   else:
    print("Perdiste")
    perdidas = perdidas + 1

historial.append(jugador + " vs " + computadora)

print("\nGanadas:", ganadas)
print("Perdidas:", perdidas)
print("Empates:", empates)

resumen = [h for h in historial]
print(resumen)
```
