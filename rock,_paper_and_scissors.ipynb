##GP1- PIEDRA, PAPEL O TIJERA
Decido hacer este quest obligatorio un poco distinto al resto, lo haré en inglés, fomentando un poco a aprender el idioma, ya que es fundamental manejar el idioma inglés en programación y en la vida.

```python
import random

opciones = ["rock", "paper", "scissors", "gun", "lightning", "devil", "dragon", "water", "air", "sponge", "wolf", "tree", "human", "snake", "fire"]

n = len(opciones)
gana_a = {}

for i in range(n):
 elemento = opciones[i]
 rivales = [opciones[(i + j) % n] for j in range(1, 8)]
 gana_a[elemento] = rivales

ganadas = 0
perdidas = 0
empates = 0
historial = []

rondas = int(input("How many rounds do you want to play? "))

for i in range(rondas):
 print("\nRound", i + 1)
 print("Options:", opciones)
 jugador = input("Choose your move: ")

 if jugador not in opciones:
  print("Invalid choice")
  continue

 computadora = random.choice(opciones)
 print("Computer chose:", computadora)

 if jugador == computadora:
  print("It's a tie!")
  empates = empates + 1
else:
  if computadora in gana_a[jugador]:
   print("You win!")
   ganadas = ganadas + 1
  else:
   print("You lose!")
   perdidas = perdidas + 1

historial.append(jugador + " vs " + computadora)

print("\nWins:", ganadas)
print("Losses:", perdidas)
print("Ties:", empates)

resumen = [h for h in historial]
print(resumen)
```
