```python
nombre=["sofia","karen","lucila","julian","mayra"]
filtro=[]
for n in nombre:
  if n[0]=="m":filtro.append(n)
filtro
```

```python
#de lal ista recursos=[10,20,15,30,25], crear una lista onde cada cantidad se multplica por 2
```

```python
recursos = [10, 20, 15, 30, 25]

doble = []
for cantidad in recursos:
 doble.append(cantidad * 2)
print(doble)
```

```python
cubos = [n ** 3 for n in range(1, 21)]
print(cubos)
```

```python
celsius = [10, 15, 30, 35, 30, 35]

fahrenheit = [(c * 9/5) + 32 for c in celsius]
print(fahrenheit)
```

```python
personajes = ["alcaldesa", "constructor", "paladín", "bardo"]

mayus = [nombre.upper() for nombre in personajes]
print(mayus)
```

```python
objetos = ["espada", "pico", "martillo", "mapa chemcoin"]

letras = [len(objeto) for objeto in objetos]
print(letras)
```

```python
divisibles = [n for n in range(1, 101) if n % 6 == 0]
print(divisibles)
```

```python
objetos = ["oro", "piedra", "madera", "diamante", "pan", "herramienta"]

largos = [o for o in objetos if len(o) > 5]
print(largos)
```

```python
pares = [o for o in largos if len(o) % 2 == 0]
print(pares)
```

```python
divisibles = [n for n in range(1, 101) if n % 3 == 0 and n % 5 == 0]
print(divisibles)
```

```python
cantidad = [3, 15, 0, 8, 21, 2]

estado = ["disponible" if c > 0 else "agotado" for c in cantidad]
print(estado)
```

```python
numeros = list(range(1, 21))

resultado = ["par" if n % 2 == 0 else "impar" for n in numeros]
print(resultado)
```

```python
precios = [5000, 12000, 30000, 8000, 45000]

con_descuento = [p * 0.8 if p >= 10000 else p for p in precios]
print(con_descuento)
```
